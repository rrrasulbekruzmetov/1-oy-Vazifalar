# 31-SAVOL:  Agar berilgan son 5 ga bo’linsa va musbat bo’lsa unda Yaxshi, 5 ga bo’linib mavfiy bo’lsa yomon , 5 ga bo’linmay musbat bo’sa o’rtacha aks xolda juda yomon degan so’z ekranga chiqarilsin
