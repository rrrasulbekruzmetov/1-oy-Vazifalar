# pandas kutubxonasini tortib olib data setni o`qib olamiz 

```python
import pandas as pd 
```

```python
df = pd.read_csv("IMDb_movies.csv")
df.info()
```

# data preprocessingni boshlaymiz 

```python
df.head(5)
```

```python
df['country'].fillna(df['country'].mode()[0], inplace=True)
```

```python
df['language'].fillna("english", inplace=True)
```

```python
df['director'].fillna(df['director'].mode()[0], inplace=True)
df['writer'].fillna(df['writer'].mode()[0], inplace=True)
df['production_company'].fillna(df['production_company'].mode()[0], inplace=True)
df['actors'].fillna(df['actors'].mode()[0], inplace=True)
df['description'].fillna(df['description'].mode()[0], inplace=True)
df['budget'].fillna(df['budget'].mode()[0], inplace=True)
```

```python
df.info()
```

# data set to`liq to`ldi

# endi encoding jarayonini boshlaymiz 

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
for col in df.columns:
        if df[col].dtype == 'object':
            if df[col].nunique() <= 6:
                dummies = pd.get_dummies(df[col], prefix=col, dtype=int)
                df = pd.concat([df.drop(columns=[col]), dummies], axis=1)
            else:
                df[col] = encoder.fit_transform(df[col]) 


# CODE ISHLAMADI :                 
```

# scaling jarayonini boshlaymiz  

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler,MinMaxScaler,RobustScaler
scaler_standart=StandardScaler()
scaler_minmax=MinMaxScaler()
scaler_robust=RobustScaler()
```

# STANDART SCALER 

```python
df['title']=scaler_standart.fit_transform(df[["title"]])
```

# MIN MAX SCALER 

```python
df['original_title']=scaler_minmax.fit_transform(df['original_title'])
```

# ROBUST SCALER

```python
df['duration']=scaler_robust.fit_transform(df['duration'])
```

# TARGET QIYMATNI TASHLAB YUBORAMIZ 

```python
numerical_col=df.select_dtypes(include=['number']).columns.drop('budget') 
```

```python
df[numerical_col]=scaler_standart.fit_transform(df[numerical_col])
```
