```python
import pandas as pd
df=pd.read_csv('bank_credit_scoring.csv')

```

```python
df.info()
```

# BOSH QIYMATLARNI TO`LDIRISHNI BOSHLAYMIZ 

```python
df['Рейтинг кредитной истории'].fillna(df['Рейтинг кредитной истории'].mode()[0], inplace=True)
```

# FIXED QIYMAT ORQALI TO`LDIRAMIZ 

```python
df['LV_AREA'].fillna("г. Минск", inplace=True)
```

# MODE YORDAMIDA TO`LIDRAMIZ 

```python
df['LV_SETTLEMENTNAME'].fillna(df['LV_SETTLEMENTNAME'].mode()[0], inplace=True)
```

# MEAN YORDAMIDA TOL`IDRAMIZ 

```python
df['SCORINGMARK'].fillna(df['SCORINGMARK'].mean(), inplace=True)
```

# FIXED QIYMAT BILAN TO`LIDRAMIZ 

```python
df['VELCOMSCORING'].fillna(-2.0, inplace=True)
```

# FOR SIKLI YORDAMIDA 10 TADAN KAM QIYMATLI QATORLARNI ANIQLAB OLAMIZ 

```python
for jadval in df.columns:
    if df[jadval].nunique() <=10:
        print(jadval)
```

#  EDU VA SEX USTUNINI ONE -HOT ENCODING BILAN TO`LDIRAMIZ 

```python
encoded = pd.get_dummies(df['SEX'], prefix="Jinsi", dtype=int)
df = pd.concat([df.drop("SEX", axis = 1), encoded], axis=1,)
```

```python
encoded = pd.get_dummies(df['EDU'], prefix="VALUE", dtype=int)
df = pd.concat([df.drop("EDU", axis = 1), encoded], axis=1,)
```

# BIRTHDATE USTUNINI LABEL ENCODING BILAN TO`LDIRAMIZ  

```python
from sklearn.preprocessing import LabelEncoder
encoded = LabelEncoder()
df['BIRTHDATE'] = encoded.fit_transform(df['BIRTHDATE'])
```

# Рейтинг кредитной истории USTUNINI FREQUENCY ENCODING QILAMIZ 

```python
freq=df['Рейтинг кредитной истории'].value_counts()
df['Рейтинг кредитной истории' + "freq"]=df['Рейтинг кредитной истории'].map(freq) 
```

# LV_AREA USTUNINI TARGET ENCODING QILAMIZ 

```python
mean_target=df.groupby('LV_AREA')['TERM'].mean()
df['LV_AREA' + "freq"]=df['LV_AREA'].map(mean_target)
```

# FOR LOOP + ONE HOT ENCODING TAXMINIY MISOLLARDA 

```python
for col in df.columns:
    if df[col].dtype=='object':
        dummies=pd.get_dummies(df[col],prefix='col', dtype=int)
        df=pd.concat([df.drop(columns=[col]),dummies],axis=1)
```

# FOR LOOP + LABEL ENCODING TAXMINIY MISOLLARDA 

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
for col in df.columns:
    if df[col].dtype=='object':
        df[col]=encoder.fit_transform(df[col])
```

# FOR LOOP + ONE HOT ENCODING VA LABEL ENCODING BIRGALIKDA 

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()

for col in df.columns:
    if df[col].dtype=='object':
        if df[col].nunique()<=6:
            dummies=pd.get_dummies(df[col],prefix='col', dtype=int)
            df=pd.concat([df.drop(columns=[col]),dummies],axis=1)

        else:
            df[col]=encoder.fit_transform(df[col])
```

# FREQUENCY ENCODING TAXMINIY MISOLLARDA 

```python
freq=df['USTUN_NOMI'].value_counts()
df['USTUN_NOMI'+'_freq']=df['USTUN_NOMI'].map(freq)
```

# TARGET ENCODING TAXMINIY MISOLLARDA 

```python
mean_target=df.groupby('USTUN_NOMI')['2_USTUN_NOMI'].mean()
df['USTUN_NOMI'+'_freq']=df['USTUN_NOMI'].map(mean_target)
```
