# Encoding Jarayonini boshlaymiz 

# ONE-HOT encoding

# Identity // Gender // Marital Status ustunlarida kam qiymat bo`lgani uchun ushbu ustunlarni ONE - HOT ENCODING orqali to`lidramiz 

```python
Identity_encoded=pd.get_dummies(df['Identity'],prefix='Qiymat',dtype=int)
df=pd.concat([df.drop('Identity',axis=1),Identity_encoded],axis=1)

Gender_encoded=pd.get_dummies(df['Gender'],prefix='Qiymat',dtype=int)
df=pd.concat([df.drop('Gender',axis=1),Gender_encoded],axis=1)

Marital_Status_encoded=pd.get_dummies(df['Marital Status'],prefix='Xolati',dtype=int)
df=pd.concat([df.drop('Marital Status',axis=1),Marital_Status_encoded],axis=1)
```

# Qolgan Ustunlarni Label Encoding orqali to`ldirib chiqamiz 

```python
encoder=LabelEncoder()
```

```python
df['PageID']=encoder.fit_transform(df['PageID'])
df['Name']=encoder.fit_transform(df['Name'])
df['Universe']=encoder.fit_transform(df['Universe'])
df['URL']=encoder.fit_transform(df['URL'])
df['Teams']=encoder.fit_transform(df['Teams'])
df['Creators']=encoder.fit_transform(df['Creators'])
```
