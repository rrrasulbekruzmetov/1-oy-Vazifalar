```python
import pandas as pd 
df=pd.read_csv('automobile_data.csv')
```

```python
df.info()
```

# MISSING VALUE 

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype=='number':
            df[col].fillna(df[col].mean(),inplace=True)
        else:
            df[col].fillna(df[col].mode()[0],inplace=True)
```

```python
df.info()
```

# ENCODING 

```python
from sklearn.preprocessing import LabelEncoder
encodlash=LabelEncoder()
```

```python
for col in df.columns:
    if df[col].dtype=='object':
        df[col]=encodlash.fit_transform(df[col])
```

```python
df.info()
```

# Scaling  jarayonini boshlaymiz 

```python
#Scaling
from sklearn.preprocessing import StandardScaler
scaling_step=StandardScaler()
```

# raqamli ustunlarni aniqlaymiz  // target qiymatni tushurib yuboramiz 

```python
num_col=df.select_dtypes(include=['int64','int32','float64']).columns.drop('price')
```

```python
num_col
```

```python
df[num_col]=scaling_step.fit_transform(df[num_col])
```

# input va output qiymatlarni belgilab olamiz 

```python
x=df.drop('price',axis=1)
```

```python
x.info()
```

# bizda price ustuni target qiymat 


```python
y=df['price'] 
```

```python
from sklearn.model_selection import train_test_split
```

# Data setni Train Test Validation qismlariga ajratib olamiz 

```python
x_train,x_temp,y_train,y_temp=train_test_split(x,y,test_size=0.2,random_state=42)
x_val,x_test,y_val,y_test=train_test_split(x_temp,y_temp,test_size=0.5,random_state=42)
```

```python
x_train
```

# Model tanlab olamiz bizda [Decision Tree Classifier] modeli tanlandi  

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model=DecisionTreeClassifier()
```

```python
model
```

# Train qilish jarayoni 

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```
