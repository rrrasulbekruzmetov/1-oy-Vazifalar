```python
import pandas as pd
```

# DATASETNI YUKLAB OLAMIZ 

```python
df = pd.read_csv("personality_dataset.csv")
```

```python
df.head()
```

# TUSHIRIB QOLDIRILGAN QIYMATLARNI TO`LDIRAMIZ 

```python
for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='object':
                df[col].fillna(df[col].mode()[0],inplace=True)
            else:
                df[col].fillna(df[col].mean(),inplace=True)
```

```python
df.info()
```

# ENCODING QILAMIZ .. ONE-HOT VA LABEL ENCODINGDAN FOYDALANAMIZ 

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()

for col in df.columns:
    if df[col].dtype=='object':
        if df[col].nunique()<=6:
            dummies=pd.get_dummies(df[col],prefix='col', dtype=int)
            df=pd.concat([df.drop(columns=[col]),dummies],axis=1)

        else:
            df[col]=encoder.fit_transform(df[col])
```

```python
df.info()
```

# RAQAMLI USTUNLARNI ALOHIDA QILIB OLAMIZ TARGET USTUNINI TASHLAB YUBORAMIZ 

```python
num_col=df.select_dtypes(include=['int64','float64']).columns.drop('col_Extrovert')
num_col
```

# STANDART SCALING ORQALI SCALING QILAMIZ 

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler,MinMaxScaler,RobustScaler
scaler_standart=StandardScaler()

df[num_col]=scaler_standart.fit_transform(df[num_col])
```

```python
df.head()
```

# TARGET USTUNINI TASHLAB YUBORAMIZ

```python
x=df.drop('col_Extrovert',axis=1)
x.info()
y=df['col_Extrovert'] 
y.info()
```

```python
from sklearn.model_selection import train_test_split
```

# DATASETNI KERAKLI QISMLARGA BO`LIB CHIQAMIZ 

```python
x_train,x_temp,y_train,y_temp=train_test_split(x,y,test_size=0.2,random_state=42)
x_val,x_test,y_val,y_test=train_test_split(x_temp,y_temp,test_size=0.5,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
```

# TRAIN QILAMIZ 

```python
ligistic_model = LogisticRegression()
ligistic_model
```

```python
ligistic_model.fit(x_train, y_train)
```

# PREDICT QILIB KO`RAMIZ 

```python
linear_pred=ligistic_model.predict(x_test)
linear_pred
```

```python
from sklearn.metrics import accuracy_score
```

# QANCHALIK TOGRI ISHLASHINI TEKSHIRIB KO`RAMIZ 

```python
linear_score=accuracy_score(y_test,linear_pred)
linear_score
```

# BOSHQA MODEL ORQALI HAM TEKSHIRIB KO`RAMIZ 

```python
Random_forest_model = RandomForestClassifier()
Random_forest_model
```

```python
Random_forest_model.fit(x_train, y_train)
```

```python
random_pred=Random_forest_model.predict(x_test)
random_pred
```

```python
random_score=accuracy_score(y_test,random_pred)
random_score
```
